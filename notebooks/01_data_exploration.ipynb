## Scanning dataset

```python
import os
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

# Define the path to your dataset (going up one level from the notebooks folder)
data_dir = '../data/PlantVillage'

# 1. Count the images in each class
if not os.path.exists(data_dir):
    print(f"Cannot find dataset at {data_dir}. Please check your extraction path.")
else:
    classes = sorted(os.listdir(data_dir))
    class_counts = {}
    
    for c in classes:
        class_path = os.path.join(data_dir, c)
        if os.path.isdir(class_path):
            class_counts[c] = len(os.listdir(class_path))

    # 2. Print the text summary
    print(f"Total Classes: {len(class_counts)}")
    print(f"Total Images: {sum(class_counts.values())}\n")
    for disease, count in class_counts.items():
        print(f"{disease}: {count} images")

    # 3. Plot the distribution chart
    plt.figure(figsize=(14, 7))
    bars = plt.bar(class_counts.keys(), class_counts.values(), color='skyblue')
    
    # Add exact numbers on top of the bars
    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, yval + 10, int(yval), ha='center', va='bottom', fontsize=8)

    plt.xticks(rotation=80)
    plt.title('PlantVillage Dataset: Class Distribution')
    plt.xlabel('Plant & Disease Class')
    plt.ylabel('Number of Images')
    plt.tight_layout()
    plt.show()
```
